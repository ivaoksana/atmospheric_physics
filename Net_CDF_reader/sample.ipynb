```python
def test_decode_time():

    calendars = [
        "standard",
        "noleap",
        "360_day",
        "all_leap"
    ]

    for calendar in calendars:

        ds = xr.Dataset(
            data_vars={
                "temperature": (
                    "time",
                    np.arange(5, dtype=float)
                )
            },
            coords={
                "time": (
                    "time",
                    np.arange(5, dtype=float),
                    {
                        "units": "days since 2000-01-01 00:00:00",
                        "calendar": calendar
                    }
                )
            }
        )

        decoded = decode_time(ds, debug=False)

        print(f"\nCalendar: {calendar}")
        print(decoded.time.values)
        print("Type:", type(decoded.time.values[0]))

test_decode_time()
```
